### 1. Criar a tabela

* Nome: alunas.

* Colunas obrigatórias:

- id (inteiro, chave primária, autoincremento)

- nome (texto, obrigatório)

- email (texto, único, obrigatório)

- data_nascimento (data, obrigatório)

- data_cadastro (data/hora, com valor padrão atual).

```python
import sqlite3
con = sqlite3.connect("cadastro.db")
cur = con.cursor()
```

```python
cur.execute("""
CREATE TABLE IF NOT EXISTS alunas (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    nome TEXT NOT NULL,
    email TEXT UNIQUE NOT NULL,
    data_nascimento DATE NOT NULL,
    data_cadastro TIMESTAMP DEFAULT CURRENT_TIMESTAMP
);
""")

con.commit()
```

#### 4. Atualizar registros

* Alterar o email de pelo menos uma aluna.

* Atualizar o nome de uma aluna existente.

* Corrigir a data de nascimento de pelo menos uma aluna.

```python
cur.execute("UPDATE alunas SET email = 'ana.souza004@example.com'  WHERE id = 1")
con.commit()

print("\nEmail atualizado aluna id = 1:")
for row in cur.execute("SELECT * FROM alunas"):
    print(row)
```

```python
cur.execute("UPDATE alunas SET nome = 'Beatriz Lima Ferraz'  WHERE id = 2")
con.commit()

print("\nNome atualizado aluna id = 2:")
for row in cur.execute("SELECT * FROM alunas"):
    print(row)
```

```python
updates = [
    (5, '1988-05-10'),
    (10, '1980-08-19'),
    (25, '1989-01-29'),
    (30, '1982-07-09')
]

for id_aluna, data in updates:
    cur.execute("UPDATE alunas SET data_nascimento = ? WHERE id = ?", (data, id_aluna))

con.commit()

print("\nDatas de nascimento atualizadas (id = 5, 10, 25, 30):")
for row in cur.execute("SELECT * FROM alunas"):
    print(row)
```

### 5. Remover registros

* Excluir pelo menos uma aluna da tabela utilizando DELETE.

* Excluir todas as alunas que tenham data de nascimento anterior a 1990.

```python
delete_ids = [
    (3,),
    (4,)
]

for id_aluna in delete_ids:
    cur.execute("DELETE FROM alunas WHERE id = ?", id_aluna)

con.commit()

print("\nDeletando alunas (id = 3,4):")
for row in cur.execute("SELECT * FROM alunas"):
    print(row)
```

```python
cur.execute("DELETE FROM alunas WHERE data_nascimento <='1990-01-01'")
con.commit()

print("\nDeletando alunas nascidas antes de 1990:")
for row in cur.execute("SELECT * FROM alunas"):
    print(row)
```
